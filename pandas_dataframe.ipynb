Filtering a DataFrame

```python
ipl.head(3)
```

```python
# find all the final winners of ipl matches
task = ipl['MatchNumber'] == 'Final'
new_df = ipl[task]
new_df[['Season','WinningTeam']]

ipl[ipl['MatchNumber'] == 'Final'][['Season','WinningTeam']]
```

```python
# how many super over matches have been played in ipl
ipl[ipl['SuperOver'] == 'Y'].shape[0]
```

```python
# how many mathes csk win in kolkata
ipl[(ipl['City'] == 'Kolkata') & (ipl['WinningTeam'] == 'Chennai Super Kings')].shape[0]
```

```python
# toss winner  is matches where the toss winner is also the match winner
ipl[ipl['TossWinner'] == ipl['WinningTeam']].shape[0]/ipl.shape[0]*100
```

```python
# movies with rating higher than 8 and votes >10000
movies[(movies['imdb_rating'] > 8) & (movies['imdb_votes'] > 10000)].shape[0]
```
